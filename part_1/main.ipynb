### 11 Skapa en ny kolumn som är heter pledged_real_diff som representerar skillnaden i USD mellan pledged usd och usd_pledged_result.
Hint: Du kommer behöva "cleana" "pledged usd" så att du kan utföra matematiska operationer på det genom att först städa bort onödig text och sedan konvertera det till en float.

```python
#df["pledged_real_diff"]
#df.info()
#df
#df["usd_pledged_result"] - df["usd pledged"]
#df["usd pledged"].sample(50)
#df["usd_pledged_result"].sample(50)

# Step one - check column information - does the columns have same datatype? NO
# usd pledged got 2345098$  got("$" "usd" "space") and is a (object)
# usd_pledged_result got  29348.00 (float64)

# Clean data - convert object to float (easier to do math operations)
# remove $ and usd with .replace and regex pattern [\$,usd ] <- removes dollar sign, "usd" and any spaces. and replace with "". then convert to float with .astype(float)


#Prefix your string with an r to tell Python not to treat backslashes specially:
df["usd pledged"] = df["usd pledged"].replace(r'[\$,usd ]', '', regex=True).astype(float)

# Sanity check - works!
df["usd pledged"].sample(10)
df["usd_pledged_result"].sample(10)

# Step two - do maths..
# Take small - big number. Else the maths gets incorrect.

df["pledged_real_diff"] = df["usd_pledged_result"] - df["usd pledged"]
df
```
