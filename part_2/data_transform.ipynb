### Enrichment: Taking the YH Points column in considuration, it would be be able to see how long it would take to achive the YH points in the cells.

Each program’s duration can be estimated using a common rule from MYH:

5 YH-points ≈ 1 study week.
So, for example, a 400 YH-point program lasts about 80 weeks (400 ÷ 5). Using this rule we can calculate the duration in weeks and years.

```python
# Using a for-loop to go through the datasets.
# Calculate study duration in weeks (5 YH points = 1 study week), and save it to the new made column. Also rounding weeks to whole numbers (int64).
# Convert weeks to years (about 52 weeks per year), and save it to the new made column. Also rounding years to .2 decimals (float).
for df in [df_2020, df_2021, df_2022]:
    df["Study Weeks"] = (df["YH Points"] / 5).round().astype("Int64")
    df["Study Years"] = (df["Study Weeks"] / 52).round(2)
```

# Harmonization:

We're gonna harmonize column order using the 2020 as the template and then concatinate them.

```python
# We create variable that holds the 2020 columns, add those to the 2021 datasets columns, and add those columns to the 2022 dataset columns.
cols = df_2020.columns
df_2021 = df_2021[cols]
df_2022 = df_2022[cols]

# We create a variable that stores the result of the concatination.
# We use the pd.concat() method that combines our DataFrames.
# We use the ignore_index=True to ignore the original index vaules from each DataFrame and create new indexes for the new combined DataFrame - else we'll get duplicates or conflicting rows indices.
combined = pd.concat([df_2020, df_2021, df_2022], ignore_index=True)

# Check the .shape - which returns the number of rows and number of columns as a tuple
print(combined.shape) # output: (1330, 11)
combined.head()


# Check if there's any duplicates and sum them up
print("Duplicate Case Numbers:", combined["Case Number"].duplicated().sum()) # output: Duplicate Case Numbers: 0
```

```python
# Check .sample(50) method to see 50 random rows of the combined dataset looks like (we can run the cell serveral times to check if there's any anomalys).
combined.sample(50)
```

```python
# Quick sanity check: Is there any Nan-values? - No.
print(combined.isna().sum())
```

### We got our 11 columns, all the cells in the columns looks fine - no obvious anomalys, index labels also looks fine and there's no duplicates or Nan-values.

```python
# Save the combined DataFrames to a .csv-file.
# Using the index=False here aswell, else we're gonna end up with a default indexs in the first column (.csv files creates them, index=False removes them).
combined.to_csv("harmonized_yh_2020_2022.csv", index=False)
```
